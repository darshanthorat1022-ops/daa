```python
def f(x):
    return (x + 3)**2
```

```python
def gradient(x):
    return 2 * (x + 3)
```

```python

x = 2
lr = 0.1
iterations = 100
```

```python
for i in range(iterations):
    x = x - lr * gradient(x)
```

```python

print("Minimum point x =", round(x, 4))
print("Minimum value y =", round(f(x), 4))
```
